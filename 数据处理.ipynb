```python
# 对所有表统一查看空缺数量与比例
def missing_summary(df, name):
    missing = df.isnull().sum()
    ratio = missing / len(df)
    result = pd.DataFrame({
        "空缺数量": missing,
        "空缺比例": ratio
    })
    result = result[result["空缺数量"] > 0].sort_values("空缺数量", ascending=False)
    print(f"\n===== {name} =====")
    print(result if not result.empty else "无空缺")
    return result
```

```python
missing_summary(orders, "orders")
missing_summary(order_items, "order_items")
missing_summary(order_payments, "order_payments")
missing_summary(order_reviews, "order_reviews")
missing_summary(customers, "customers")
missing_summary(products, "products")
missing_summary(sellers, "sellers")
missing_summary(product_category_name_translation, "product_category_name_translation")
```

### 数据预处理

```python
import os
os.makedirs('data', exist_ok=True)

# 订单表：日期列转 datetime（缺失为未支付/未发货/未送达，业务合理，保留不填充）
for col in ['order_purchase_timestamp', 'order_approved_at',
            'order_delivered_carrier_date', 'order_delivered_customer_date',
            'order_estimated_delivery_date']:
    orders[col] = pd.to_datetime(orders[col], errors='coerce')

# 评价表：评论标题/内容缺失 88%/59%，只保留有用列
order_reviews = order_reviews[['review_id', 'order_id', 'review_score',
                               'review_creation_date', 'review_answer_timestamp']]

# 商品表：品类缺失标记为 unknown，并补充英文翻译
products['product_category_name'] = products['product_category_name'].fillna('unknown')
products = products.merge(product_category_name_translation, on='product_category_name', how='left')
products['product_category_name_english'] = products['product_category_name_english'].fillna('other')

# 地理位置表：去重（26 万重复行）
geolocation = geolocation.drop_duplicates()

# 保存清洗后的所有表
cleaned = {'customers': customers, 'geolocation': geolocation,
           'order_items': order_items, 'order_payments': order_payments,
           'order_reviews': order_reviews, 'orders': orders,
           'products': products, 'sellers': sellers}
for name, df in cleaned.items():
    df.to_csv(f'data/{name}_clean.csv', index=False, encoding='utf-8-sig')
print('已保存:', os.listdir('data'))
```

## 业务解读

- 清洗阶段决定了所有下游分析的口径：日期字段转 datetime 是时间序列分析的前提；品类缺失标为 unknown、翻译缺失标为 other，保留了这些订单而不是丢弃，避免样本偏差。
- reviews 只保留评分与时间列——Olist 评论文字缺失高达 88%，评论价值密度低，但 1–5 分评分覆盖率接近 100%，是更可靠的满意度信号。
- geolocation 去重后仍有 73.8 万行，且经纬度只精确到邮编前缀级别，做地图分析时按邮编聚合使用，不要按行读取。

```python
import matplotlib.pyplot as plt
plt.rcParams['font.sans-serif'] = ['Microsoft YaHei']
plt.rcParams['axes.unicode_minus'] = False

src_orders = pd.read_csv('源数据/olist_orders_dataset.csv', parse_dates=['order_purchase_timestamp'])

fig, axes = plt.subplots(1, 2, figsize=(13.5, 5))

# ① 订单状态分布饼图
st = src_orders['order_status'].value_counts()
axes[0].pie(st.values, labels=st.index, autopct='%1.1f%%',
            startangle=90, textprops={'fontsize': 9})
axes[0].set_title(f'订单状态分布（共 {len(src_orders):,} 单）')

# ② 月度订单量折线（确认 2016 残缺、2018-09 截断两个边界）
g = src_orders['order_purchase_timestamp'].dt.to_period('M').value_counts().sort_index()
xlab = [str(i) for i in g.index]
axes[1].plot(range(len(g)), g.values, color='#3498db', marker='o', ms=3)
axes[1].set_xticks(range(0, len(g), 3), xlab[::3], rotation=45)
axes[1].set_xlabel('下单月份'); axes[1].set_ylabel('订单数')
axes[1].set_title('月度订单量折线（源数据粗查）')

plt.tight_layout()
plt.show()
```
